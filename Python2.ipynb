```python
#if-elif-else statement
Age = int(input("enter your Age:"))
if Age >= 18:
    print("you are adult eligible to vote")
elif Age <= 10:
    print("you are small kid cant vote")
else:
    print("you are teenager cant vote")
```

```python
#if-else statement
Marks = int(input("enter your percentage"))
if Marks >=34:
    print("you are pass")
else:
    print("you are fail")
```

```python
#if statement
Year = int(input("enter your birth year:"))
if Year >= 2010:
    print("you are Gen Alpha")
```

```python
#nested if statement
age = int(input("enter your age:"))
if age >= 18:
    print("you are adult")
    if age >= 60:
        print("you are senior citizen")
    else:
        print("you are not senior citizen")
else:
    print("you are not adult")
```

```python
# while loop
i = 1
while i<=5:
    print(i)
    i+=1
```

```python
# for loop statement
for i in range(5):
    print(i)
```

```python
# Nested for loop
for i in range(2):
    for j in range(2):
        print(i,j)
```

```python
# break statement
for i in range(1,6):
    if i == 5:
        break
    print(i)
```

```python
# continue statement
for i in range(1,6):
    if i == 4:
        continue
    print(i)
```

```python
# pass statement
for i in range(1,6):
    if i == 3:
         pass
    print(i)
```

```python
#arthmetic operators
x = 10
y = 5

print("addition:", x + y)
print("subtraction:", x - y)
print("multiplication:", x * y)
print("division:", x / y)
print("modulus:", x % y)
```

```python
# Relational operators
a = 10
b = 5
print("a > b:",a > b)
print("a < b:",a < b)
print("a == b:",a == b)
print("a != b:",a != b)
print("a >= b:",a >= b)
print("a <= b:",a <= b)
```

```python
#logical operators
a = 10
b = 5

print(a > 5 and b < 10)
print(a > 15 or b < 10)
print(not(a > 5))
```
